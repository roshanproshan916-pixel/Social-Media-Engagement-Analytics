# Task 1: Data Import and Setup
# Importing required Python libraries

```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

# Display settings
pd.set_option('display.max_columns', None)

print("Libraries imported successfully!")
```

## 1.2 Loading the Dataset

```python
# Load the social media engagement dataset
df = pd.read_csv("social_media_engagement_5000.csv")

# Display first five rows
df.head()
```

## 1.3 Checking Dataset Structure

```python
# Display first five rows
print("First 5 Rows:")
display(df.head())

# Display last five rows
print("Last 5 Rows:")
display(df.tail())

# Check dataset shape
print("Dataset Shape:", df.shape)

# Display column names
print("Column Names:")
print(df.columns)
```

## 1.4 Checking Data Types and Information

```python
# Display dataset information
df.info()

# Check data types
print("\nData Types:")
print(df.dtypes)
```

## 1.5 Converting Date Column into Datetime

```python
# Convert posted_at column into datetime format
df['posted_at'] = pd.to_datetime(df['posted_at'], dayfirst=True)

# Check updated data type
print(df['posted_at'].dtype)

# Display first five dates
df[['posted_at']].head()
```

## 1.6 Basic NumPy Array Operations

```python
# Convert likes column into NumPy array
likes_array = np.array(df['likes'].dropna())

# Display array
print("Likes Array:")
print(likes_array[:10])

# Calculate sum
print("Total Likes:", np.sum(likes_array))

# Calculate average
print("Average Likes:", np.mean(likes_array))

# Find maximum and minimum likes
print("Maximum Likes:", np.max(likes_array))
print("Minimum Likes:", np.min(likes_array))
```

## Task 2: Data Cleaning
### 2.1 Detecting Missing Values

```python
# Check missing values in each column
print("Missing Values:")
print(df.isnull().sum())

# Check total missing values
print("\nTotal Missing Values:")
print(df.isnull().sum().sum())
```

### 2.2 Missing Values Percentage

```python
# Calculate missing values percentage
missing_percentage = (df.isnull().sum() / len(df)) * 100

print("Missing Values Percentage:")
print(missing_percentage)
```

### 2.3 Handling Missing Values

```python
# Fill numerical missing values with median
for col in df.select_dtypes(include=np.number).columns:
    df[col] = df[col].fillna(df[col].median())

# Fill categorical missing values with mode
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].fillna(df[col].mode()[0])

print("Missing values handled successfully!")
```

### 2.4 Verifying Missing Values

```python
# Check missing values after cleaning
print(df.isnull().sum())

print("\nTotal Missing Values:", df.isnull().sum().sum())
```

### 2.5 Identifying and Removing Duplicates

```python
# Check duplicate records
print("Number of Duplicate Rows:", df.duplicated().sum())

# Remove duplicate records
df.drop_duplicates(inplace=True)

# Check dataset shape after removing duplicates
print("Dataset Shape:", df.shape)
```

### 2.6 Standardizing Categories

```python
# Standardize gender values
df['gender'] = df['gender'].str.strip().str.title()

# Standardize sentiment values
df['sentiment'] = df['sentiment'].str.strip().str.lower()

# Display unique values
print("Gender Categories:")
print(df['gender'].unique())

print("\nSentiment Categories:")
print(df['sentiment'].unique())
```

### 2.7 Checking and Correcting Unrealistic Values

```python
# Check negative values
columns = ['likes', 'comments', 'shares']

for col in columns:
    print(col, "Negative Values:", (df[col] < 0).sum())

# Replace negative values with median
for col in columns:
    df.loc[df[col] < 0, col] = df[col].median()

print("\nUnrealistic values checked and corrected!")
```

### 2.8 Extracting Hashtag Count

```python
# Count hashtags in each post
df['hashtag_count'] = df['hashtags'].str.count('#')

# Display hashtag information
display(df[['hashtags', 'hashtag_count']].head(10))
```

### 2.9 Final Data Cleaning Verification

```python
# Check final dataset information
print("Final Dataset Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nFinal Dataset:")
display(df.head())
```

# Task 4: Data Wrangling
## 4.1 Combining DataFrames using Concat

```python
# Split dataset into two parts
df1 = df.iloc[:2500]
df2 = df.iloc[2500:]

# Combine both DataFrames
combined_df = pd.concat([df1, df2], ignore_index=True)

# Display combined dataset
print("Combined Dataset Shape:", combined_df.shape)
display(combined_df.head())
```

## 4.2 Creating Engagement Score

```python
# Apply log transformation to numerical metrics
df['log_likes'] = np.log1p(df['likes'])
df['log_comments'] = np.log1p(df['comments'])
df['log_shares'] = np.log1p(df['shares'])

# Display transformed values
display(df[['likes', 'log_likes',
            'comments', 'log_comments',
            'shares', 'log_shares']].head(10))
```

# Task 6: Data Visualization
## Part A: Matplotlib
### 6.1 Scatter Plot - Likes vs Impressions

```python
# Create scatter plot
plt.figure(figsize=(8, 5))

plt.scatter(df['likes'], df['impression_count'], alpha=0.5)

plt.title('Likes vs Impressions')
plt.xlabel('Likes')
plt.ylabel('Impressions')
plt.grid(True)

plt.show()
```

### 6.2 Line Plot - Daily Engagement Trend

```python
# Convert posted_at to datetime
df['posted_at'] = pd.to_datetime(df['posted_at'], dayfirst=True)

# Calculate daily engagement
daily_engagement = df.groupby(
    df['posted_at'].dt.date
)['engagement_rate'].mean()

# Create line plot
plt.figure(figsize=(12, 5))

plt.plot(daily_engagement.index, daily_engagement.values)

plt.title('Daily Engagement Trend')
plt.xlabel('Date')
plt.ylabel('Average Engagement Rate')
plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()
```

### 6.3 Bar Plot - Posts by Category
